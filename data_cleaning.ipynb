```python
import pandas as pd
from google.colab import files
uploaded = files.upload() # Une fenêtre s'ouvrira pour sélectionner le fichier sur votre PC

df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
# 2. Aperçu des données et des types
print("--- Structure du dataset ---")
print(df.info())

print("\n--- Aperçu des 5 premières lignes ---")
print(df.head())
```

```python
# Convert TotalCharges to numeric (spaces become NaN)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Check created missing values
print("Number of missing values in TotalCharges:", df['TotalCharges'].isnull().sum())

# Replace NaNs with 0 (or with MonthlyCharges value if tenure == 0)
df['TotalCharges'] = df['TotalCharges'].fillna(0)

# Check for duplicates
print("Number of duplicates:", df.duplicated().sum())
```

```python
# Calculate KPIs
total_customers = len(df)
churn_rate = (df['Churn'] == 'Yes').mean() * 100
avg_monthly_charges = df['MonthlyCharges'].mean()

print(f"Total Customers: {total_customers}")
print(f"Churn Rate: {churn_rate:.2f}%")
print(f"Average Monthly Charge: {avg_monthly_charges:.2f} $")
```

```python
# Save cleaned dataset to a new CSV file
df.to_csv('cleaned_telco_churn.csv', index=False)
print("File 'cleaned_telco_churn.csv' successfully saved!")
```
